## Exceptions

If things go wrong during the execution of the program(runtime). It generally happens when something unforeseen has happened.

- Exceptions are raised by python runtime
- You have to takle is on the fly

#### **Examples**

- Memory overflow
- Divide by 0 -> logical error
- Database error

```python
# Why is it important to handle exceptions

# i) to make better UX by not showing them stacktrace.
# ii ) to avoid hackers and loop holes to prevent security breaches.
```

```python
with open('file.txt' , 'w') as f:
  f.write('python language')
```

```python
# try except

# Whenever you are working with external sources like files , DB connections its a risky piece of code and must be handled.

try:
  with open('file1.txt' , 'r') as f:
    print(f.read())
except:
  print("Sorry! File Not Found.")
```

### Catching Specific Exceptions

```python

# Best practice is to handle each exception and put a generic one at the end. Generic box must be at last otherwise it will take over other exception blocks.

try:
  m=5
  f = open("file.txt" , 'r')
  print(f.read())
  print(m)
  print(5/2)
  L = [12,12,45]
  L[10]
except FileNotFoundError:
  print("File Not Found.")
except NameError:
  print("Variable not Defined.")
except ZeroDivisionError:
  print('cannot divide by 0')
except Exception as e:
  print(e)
```

###  `else` Block

```python

# Put code to be executed after success of try in else block. This contains code that we are super sure that its correct.
try:
  f = open('file.txt','r')
except FileNotFoundError:
  print("File Not Found.")
except Exception as e:
  print(e)
else:
  print(f.read())
```

### `finally` block

```python
 # Code in this will be executed in every situation. Used to close connections like socket , bluetooth , files etc.  This will prevent security breaches.

try:
  f = open('file.txt', 'r')
except FileNotFoundError:
  print("File Not Found.")
except Exception as e:
  print(e)
else:
  print(f.read())
finally:
  f.close()
  print("File closed.")
```

```python
class Bank:

  def __init__(self, balance):
    self.balance = balance

  def withdraw(self,amount):
    if amount < 0:
      raise Exception("Amount Cannot be Negitive.")
    elif self.balance < amount:
      raise Exception("Balance is less than Required Amount.")
    self.balance = self.balance - amount
    print(f'{amount} Debited from your Account.')


obj = Bank(10000)

try:
  obj.withdraw(15000)
except Exception as e:
  print(e)
```

### Custom Exceptions

#### Why to have custom class?
i ) App based errors

ii) full control

iii ) for custom code logics


```python
class MyExceptionClass(Exception):  # Custom exception class must be inherited from Exception class.
  def __init__(self, message):
    print(message)

class Bank:

  def __init__(self, balance):
    self.balance = balance

  def withdraw(self,amount):
    if amount < 0:
      raise MyExceptionClass("Amount Cannot be Negitive.")
    elif self.balance < amount:
      raise MyExceptionClass("Balance is less than Required Amount.")
    self.balance = self.balance - amount
    print(f'{amount} Debited from your Account.')


obj = Bank(10000)

try:
  obj.withdraw(5000)
except MyExceptionClass as e:
  print(e)
```

```python
class SecurityError(Exception):
  def __init__(self, message):
    print(message)

  def logout(self):
    print("Logging Out From all Devices.")


class Google:
  def __init__(self ,name, gmail , password , device):
    self.name = name
    self.gmail = gmail
    self.password = password
    self.device = device

    # device is detected by some code logic.


  def login(self,gmail , password , device):
    if device != self.device:
      raise SecurityError("Unauthorized Device Detected!")

    if gmail == self.gmail and password == self.password:
      print("Login Successfull.")
    else:
      print("Login Error.")



obj = Google('Sam' , 'sam@gmail.com' , '123456', 'android')

try:
  obj.login('sam@gmail.com' , '123455', 'ios')
except SecurityError as e:
  e.logout()
else:
  print(obj.name)
finally:
  print("DB Connection Closed.")
```
